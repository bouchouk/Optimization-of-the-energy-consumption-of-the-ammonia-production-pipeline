```python

#define the population size and number of generations
POPULATION_SIZE = 100
NUM_GENERATIONS = 500
MUTATION_RATE = 0.5
ELITISM_SIZE = 20  # Number of top individuals to carry over


# Define parameter ranges
ranges_dict = {col: {'Min': df[col].min(), 'Max': df[col].max()} for col in df.columns}
ranges_dict['NG FEED - Comp Mass Flow (KJ/h)']={'Min':50,'Max':100}
ranges_dict['Ratio (S/C)']={'Min':2.5,'Max':5}
ranges_dict['Reformer inlet temperature (°C)']={'Min':750,'Max':900}
ranges_dict['Reformer inlet pressure (kPa)']={'Min':2000,'Max':3500}
ranges_dict['Reformer outlet temperature (°C)']={'Min':800,'Max':1000}
ranges_dict['HTS - Inlet temperature (°C)']={'Min':300,'Max':450}
ranges_dict['HTS - Inlet pressure (kPa)']={'Min':1500,'Max':3500}
ranges_dict['LTS - Inlet temperature (°C)']={'Min':200,'Max':300}
ranges_dict['LTS - Inlet pressure (kPa)']={'Min':1500,'Max':3000}
ranges_dict['Condensor inlet temperature (°C)']={'Min':100,'Max':212.8}
ranges_dict['Condensor inlet pressure (kPa)']={'Min':800,'Max':2000}
ranges_dict['Methanator inlet temperature (°C)']={'Min':250,'Max':400}
ranges_dict['Hydrogen inlet temperature (°C)']={'Min':250,'Max':450}
ranges_dict['Nitrogen inlet temperature (°C)']={'Min':250,'Max':450}
ranges_dict['Nitrogen inlet pressure (kPa)']={'Min':1000,'Max':2500}
ranges_dict['Synthesis gas inlet temperature (°C)']={'Min':350,'Max':500}
ranges_dict['Synthesis gas inlet pressure (kPa)']={'Min':15000,'Max':20000}
ranges_dict['Separator inlet temperature (°C)']={'Min':-33,'Max':10}
ranges_dict['K-100 - Delta P (kPa)']={'Min':100,'Max':400}
ranges_dict['K-102 - Delta P (kPa)']={'Min':1000,'Max':2500}


# It will helps us to go from index to name
predictures_list =['total_energy','NG FEED - Comp Mass Flow (KJ/h)','Ratio (S/C)','Reformer inlet temperature (°C)',
    'Reformer outlet temperature (°C)','HTS - Inlet temperature (°C)','LTS - Inlet temperature (°C)','Condensor inlet temperature (°C)',
    'Condensor inlet pressure (kPa)','Methanator inlet temperature (°C)','LTS - Inlet pressure (kPa)',
    'Hydrogen inlet temperature (°C)','Nitrogen inlet temperature (°C)','K-100 - Delta P (kPa)','Synthesis gas inlet temperature (°C)','Separator inlet temperature (°C)',
    'K-102 - Delta P (kPa)','Synthesis gas inlet pressure (kPa)','Separator inlet pressure (kPa)'
    ,'Ammonia - Comp Mass Flow (kg/h)']

#load the data 
    #S_models
modelS2 = joblib.load('/home/houssam/internship project/trained_models/S-2 stream temperature_Pipeline.pkl')
modelS3 = joblib.load('/home/houssam/internship project/trained_models/S-3 stream temperature_Pipeline.pkl')
    #energy
model1 = joblib.load('/home/houssam/internship project/trained_models/Mixed_Steam_Pipeline.pkl')
model2 = joblib.load('/home/houssam/internship project/trained_models/Reformer_Energy_Pipeline.pkl')
model3 = joblib.load('/home/houssam/internship project/trained_models/Q-102 - Energy_Pipeline.pkl')
model4 = joblib.load('/home/houssam/internship project/trained_models/E-1- Energy_Pipeline.pkl')
model5 = joblib.load('/home/houssam/internship project/trained_models/E-2- Energy_Pipeline.pkl')
model6 = joblib.load('/home/houssam/internship project/trained_models/E-3- Energy_Pipeline.pkl')
model7 = joblib.load('/home/houssam/internship project/trained_models/E-4- Energy_Pipeline.pkl')
model8 = joblib.load('/home/houssam/internship project/trained_models/ E-separator- Energy_Pipeline.pkl')

    #Ammonia Mass
model9 = joblib.load('/home/houssam/internship project/trained_models/ Ammonia - Comp Mass Flow_Pipeline.pkl')

    #CO2 emission
model10 = joblib.load('/home/houssam/internship project/trained_models/CO2 emissions-Comp Mass Flow_Pipeline.pkl')



# Load the trained model
'''
total_energy = X0
NG FEED - Comp Mass Flow (KJ/h) = X1
Ratio (S/C) = X2
Reformer inlet temperature (°C) = X3
Reformer outlet temperature (°C) = X4
HTS - Inlet temperature (°C) = X5
LTS - Inlet temperature (°C) = X6
Condensor inlet temperature (°C) = X7
Condensor inlet pressure (kPa) = X8
Methanator inlet temperature (°C) = X9

Mixed Steam- Comp Mass Flow (KJ/h) = model1.predict(X1,X2)
LTS - Inlet pressure (kPa) = X10
Hydrogen inlet temperature (°C) = X11
Nitrogen inlet temperature (°C) = X12
K-100 - Delta P (kPa) = X13
S-2 stream temperature (°C) = modelS2.predict(X11,X12,X13)
S-3 stream temperature (°C) = modelS3.predict(X1,X2,X10,X11,X12,X13)
Synthesis gas inlet temperature (°C) = X14
Separator inlet temperature (°C) = X15
K-102 - Delta P (kPa) = X16
Synthesis gas inlet pressure (kPa) = X17
Separator inlet pressure (kPa) = X18


'''
def predict_energy(X1,X2,X3,X4,X5,X6,X7,X8,X9,X10,X11,X12,X13,X14,X15,model1,model2,model3,model4,model5,model6,model7,model8):
    Total_energy = 0
    
    Mixed_Steam = model1.predict([[X1,X2]])[0]
    S2 = modelS2.predict([[X11,X12,X13]])[0]
    S3 = modelS3.predict([[X11,X12,X13,Mixed_Steam,X10]])[0]
    Total_energy += model2.predict([[X3,X4]])[0]
    Total_energy += 4227654.693147199*X3 - 1507124898.3797724
    Total_energy += model3.predict([[X4,X5]])[0]
    Total_energy += -3082293.5646714796*X5 + 1077098983.4823322
    Total_energy += -2640932.047069013*X6 + 953513207.3016243
    Total_energy += -2640920.0637122733*X6 + 559925580.9712701
    Total_energy += -2318202.284356667*X7 + 509146085.55179334
    Total_energy += -2318641.333250072*X7 + 20190755.60255745
    Total_energy += 11485.991953742923*X8 - 3851589.240937948
    Total_energy += 2750563.303329907*X9 - 107470876.00437343
    Total_energy += -2750561.059656047*X9 + 770251086.4382389
    Total_energy += model4.predict([[Mixed_Steam,X10,X11,X13,S2]])[0]
    Total_energy += model5.predict([[S2,S3,X13]])[0]
    Total_energy += model6.predict([[Mixed_Steam,X10,X14]])[0]
    Total_energy += -2579.93651444447*X14 + 747019.723192719
    Total_energy += model7.predict([[Mixed_Steam, X10,X15]])[0]
    Total_energy += model8.predict([[X15]])[0]

    return Total_energy

def predict_Mass(X1,X2,X10,X11,X12,X13,X16,X17,X18,model1,model9):
    Mixed_Steam = model1.predict([[X1,X2]])[0]
    S2 = modelS2.predict([[X11,X12,X13]])[0]
    S3 = modelS3.predict([[X11,X12,X13,Mixed_Steam,X10]])[0]
    output_mass = model9.predict([[Mixed_Steam,X10,S2,S3,X13,X16,X17,X18]])[0]
    return output_mass

def CO2_emission(X1,X2,X10,model1,model10):
    Mixed_Steam = model1.predict([[X1,X2]])[0]
    output = model10.predict([[Mixed_Steam,X10]])[0]
    return output



def create_population(size,predictures_list):
    population = []
    for i in range(size):
        ind = []
        for predicture in predictures_list:
            Max = ranges_dict[predicture]['Max']
            Min = ranges_dict[predicture]['Min']
            ind.append(random.uniform(Min,Max))
        population.append(ind)
    return population
    

def fitness_function(individual):
        predicted_mass = predict_Mass(individual[1],individual[2],individual[10],individual[11],individual[12],individual[13],individual[16],individual[17],individual[18],model1,model9)
        predicted_CO2_mass = CO2_emission(individual[1],individual[2],individual[10],model1,model10)
        if predicted_mass > 164 and 172 > predicted_mass and predicted_CO2_mass < 256:
            predicted_energy = predict_energy(individual[1],individual[2],individual[3],individual[4],individual[5],individual[6],individual[7],individual[8],individual[9],individual[10],individual[11],individual[12],individual[13],individual[14],individual[15],model1,model2,model3,model4,model5,model6,model7,model8)
            return -predicted_energy
        else:
            return -float('inf')


def tournament_selection(population, fitnesses, k=3):
    selected = random.sample(list(zip(population, fitnesses)), k)
    return min(selected, key=lambda item: item[1])[0]


    
def crossover(parent1, parent2):
    crossover_point = random.randint(2, len(parent1) - 2)
    child1 = parent1[:crossover_point] + parent2[crossover_point:]
    child2 = parent2[:crossover_point] + parent1[crossover_point:]
    return child1, child2



def mutate(individual):
    """Mutate an individual's parameters."""
    if random.random() < MUTATION_RATE:
        mutation_index = random.randint(1, len(individual) - 2)
        Max = ranges_dict[predictures_list[mutation_index]]['Max']
        Min = ranges_dict[predictures_list[mutation_index]]['Min']
        individual[mutation_index] = random.uniform(Min,Max)
    return individual





#the main algo 

# Create initial population (type: df)
population = create_population(POPULATION_SIZE,predictures_list)

for generation in range(NUM_GENERATIONS):
    # Evaluate fitness of each individual
    fitnesses = [fitness_function(ind) for ind in population] 


    # Sort the population by fitness and preserve the top individuals
    sorted_population = [ind for ind, fit in sorted(zip(population, fitnesses), key=lambda x: x[1], reverse=True)]
    elites = sorted_population[:ELITISM_SIZE]
    elites_fitnesses = [fitness_function(ind) for ind in elites]

    # Create a new population through selection, crossover, and mutation
    new_population = elites  # Start with the best individuals

    while len(new_population) < POPULATION_SIZE:
        parent1 = tournament_selection(elites, elites_fitnesses)
        parent2 = tournament_selection(elites, elites_fitnesses)
        child1, child2 = crossover(parent1, parent2)
        new_population.append(mutate(child1))
        if len(new_population) < POPULATION_SIZE:
            new_population.append(mutate(child2))

        
    # Replace the old population with the new population
    population = new_population

    # Print the best individual of the current generation
    best_individual = max(population, key=fitness_function)
    best_fitness = fitness_function(best_individual)
    print(f'Generation {generation}: Best Fitness = {best_fitness}, Best Individual = {best_individual}')



    
```

```python

#define the population size and number of generations
POPULATION_SIZE = 100
NUM_GENERATIONS = 500
MUTATION_RATE = 0.5
ELITISM_SIZE = 20  # Number of top individuals to carry over


# Define parameter ranges
ranges_dict = {col: {'Min': df[col].min(), 'Max': df[col].max()} for col in df.columns}



# It will helps us to go from index to name
predictures_list =['total_energy','NG FEED - Comp Mass Flow (KJ/h)','Ratio (S/C)','Reformer inlet temperature (°C)',
    'Reformer outlet temperature (°C)','HTS - Inlet temperature (°C)','LTS - Inlet temperature (°C)','Condensor inlet temperature (°C)',
    'Condensor inlet pressure (kPa)','Methanator inlet temperature (°C)','LTS - Inlet pressure (kPa)',
    'Hydrogen inlet temperature (°C)','Nitrogen inlet temperature (°C)','K-100 - Delta P (kPa)','Synthesis gas inlet temperature (°C)','Separator inlet temperature (°C)',
    'K-102 - Delta P (kPa)','Synthesis gas inlet pressure (kPa)','Separator inlet pressure (kPa)'
    ,'Ammonia - Comp Mass Flow (kg/h)']

#load the data 
    #S_models
modelS2 = joblib.load('/home/houssam/internship project/trained_models/S-2 stream temperature_Pipeline.pkl')
modelS3 = joblib.load('/home/houssam/internship project/trained_models/S-3 stream temperature_Pipeline.pkl')
    #energy
model1 = joblib.load('/home/houssam/internship project/trained_models/Mixed_Steam_Pipeline.pkl')
model2 = joblib.load('/home/houssam/internship project/trained_models/Reformer_Energy_Pipeline.pkl')
model3 = joblib.load('/home/houssam/internship project/trained_models/Q-102 - Energy_Pipeline.pkl')
model4 = joblib.load('/home/houssam/internship project/trained_models/E-1- Energy_Pipeline.pkl')
model5 = joblib.load('/home/houssam/internship project/trained_models/E-2- Energy_Pipeline.pkl')
model6 = joblib.load('/home/houssam/internship project/trained_models/E-3- Energy_Pipeline.pkl')
model7 = joblib.load('/home/houssam/internship project/trained_models/E-4- Energy_Pipeline.pkl')
model8 = joblib.load('/home/houssam/internship project/trained_models/ E-separator- Energy_Pipeline.pkl')

    #Ammonia Mass
model9 = joblib.load('/home/houssam/internship project/trained_models/ Ammonia - Comp Mass Flow_Pipeline.pkl')

    #CO2 emission
model10 = joblib.load('/home/houssam/internship project/trained_models/CO2 emissions-Comp Mass Flow_Pipeline.pkl')



# Load the trained model
'''
total_energy = X0
NG FEED - Comp Mass Flow (KJ/h) = X1
Ratio (S/C) = X2
Reformer inlet temperature (°C) = X3
Reformer outlet temperature (°C) = X4
HTS - Inlet temperature (°C) = X5
LTS - Inlet temperature (°C) = X6
Condensor inlet temperature (°C) = X7
Condensor inlet pressure (kPa) = X8
Methanator inlet temperature (°C) = X9

Mixed Steam- Comp Mass Flow (KJ/h) = model1.predict(X1,X2)
LTS - Inlet pressure (kPa) = X10
Hydrogen inlet temperature (°C) = X11
Nitrogen inlet temperature (°C) = X12
K-100 - Delta P (kPa) = X13
S-2 stream temperature (°C) = modelS2.predict(X11,X12,X13)
S-3 stream temperature (°C) = modelS3.predict(X1,X2,X10,X11,X12,X13)
Synthesis gas inlet temperature (°C) = X14
Separator inlet temperature (°C) = X15
K-102 - Delta P (kPa) = X16
Synthesis gas inlet pressure (kPa) = X17
Separator inlet pressure (kPa) = X18

'Mixed Steam- Comp Mass Flow (KJ/h)', 'LTS - Inlet pressure (kPa)', 'S-2 stream temperature (°C)', 'S-3 stream temperature (°C)',
      'K-100 - Delta P (kPa)', 'K-102 - Delta P (kPa)','Synthesis gas inlet pressure (kPa)','Separator inlet pressure (kPa)'

'''
def predict_energy(X1,X2,X3,X4,X5,X6,X7,X8,X9,X10,X11,X12,X13,X14,X15,model1,model2,model3,model4,model5,model6,model7,model8):
    Total_energy = 0
    
    Mixed_Steam = model1.predict([[X1,X2]])[0]
    S2 = modelS2.predict([[X11,X12,X13]])[0]
    S3 = modelS3.predict([[X11,X12,X13,Mixed_Steam,X10]])[0]
    Total_energy += model2.predict([[X3,X4]])[0]
    Total_energy += 4227654.693147199*X3 - 1507124898.3797724
    Total_energy += model3.predict([[X4,X5]])[0]
    Total_energy += -3082293.5646714796*X5 + 1077098983.4823322
    Total_energy += -2640932.047069013*X6 + 953513207.3016243
    Total_energy += -2640920.0637122733*X6 + 559925580.9712701
    Total_energy += -2318202.284356667*X7 + 509146085.55179334
    Total_energy += -2318641.333250072*X7 + 20190755.60255745
    Total_energy += 11485.991953742923*X8 - 3851589.240937948
    Total_energy += 2750563.303329907*X9 - 107470876.00437343
    Total_energy += -2750561.059656047*X9 + 770251086.4382389
    Total_energy += model4.predict([[Mixed_Steam,X10,X11,X13,S2]])[0]
    Total_energy += model5.predict([[S2,S3,X13]])[0]
    Total_energy += model6.predict([[Mixed_Steam,X10,X14]])[0]
    Total_energy += -2579.93651444447*X14 + 747019.723192719
    Total_energy += model7.predict([[Mixed_Steam, X10,X15]])[0]
    Total_energy += model8.predict([[X15]])[0]

    return Total_energy

def predict_Mass(X1,X2,X10,X11,X12,X13,X16,X17,X18,model1,model9):
    Mixed_Steam = model1.predict([[X1,X2]])[0]
    S2 = modelS2.predict([[X11,X12,X13]])[0]
    S3 = modelS3.predict([[X11,X12,X13,Mixed_Steam,X10]])[0]
    output_mass = model9.predict([[Mixed_Steam,X10,S2,S3,X13,X16,X17,X18]])[0]
    return output_mass

def CO2_emission(X1,X2,X10,model1,model10):
    Mixed_Steam = model1.predict([[X1,X2]])[0]
    output = model10.predict([[Mixed_Steam,X10]])[0]
    return output



def create_population(df,size,predictures_list):
    sorted_df = df.sort_values(by='total_energy').head(size)
    sorted_df = sorted_df[predictures_list]
    return sorted_df.values.tolist()
    


def fitness_function(individual):
        predicted_mass = predict_Mass(individual[1],individual[2],individual[10],individual[11],individual[12],individual[13],individual[16],individual[17],individual[18],model1,model9)
        predicted_CO2_mass = CO2_emission(individual[1],individual[2],individual[10],model1,model10)
        if predicted_mass > 164 and 172 > predicted_mass and predicted_CO2_mass < 256:
            predicted_energy = predict_energy(individual[1],individual[2],individual[3],individual[4],individual[5],individual[6],individual[7],individual[8],individual[9],individual[10],individual[11],individual[12],individual[13],individual[14],individual[15],model1,model2,model3,model4,model5,model6,model7,model8)
            return -predicted_energy
        else:
            return -float('inf')


def tournament_selection(population, fitnesses, k=3):
    selected = random.sample(list(zip(population, fitnesses)), k)
    return min(selected, key=lambda item: item[1])[0]


    
def crossover(parent1, parent2):
    crossover_point = random.randint(2, len(parent1) - 2)
    child1 = parent1[:crossover_point] + parent2[crossover_point:]
    child2 = parent2[:crossover_point] + parent1[crossover_point:]
    return child1, child2



def mutate(individual):
    """Mutate an individual's parameters."""
    if random.random() < MUTATION_RATE:
        mutation_index = random.randint(1, len(individual) - 2)
        Max = ranges_dict[predictures_list[mutation_index]]['Max']
        Min = ranges_dict[predictures_list[mutation_index]]['Min']
        individual[mutation_index] = random.uniform(Min,Max)
    return individual





#the main algo 

# Create initial population (type: df)
population = create_population(df,POPULATION_SIZE,predictures_list)
fitnesses = [-ind[0] for ind in population]
# Sort the population by fitness and preserve the top individuals
sorted_population = [ind for ind, fit in sorted(zip(population, fitnesses), key=lambda x: x[1], reverse=True)]
elites = sorted_population[:ELITISM_SIZE]
elites_fitnesses = [fitness_function(ind) for ind in elites]

# Create a new population through selection, crossover, and mutation
new_population = elites  # Start with the best individuals

while len(new_population) < POPULATION_SIZE:
    parent1 = tournament_selection(elites, elites_fitnesses)
    parent2 = tournament_selection(elites, elites_fitnesses)
    child1, child2 = crossover(parent1, parent2)
    new_population.append(mutate(child1))
    if len(new_population) < POPULATION_SIZE:
        new_population.append(mutate(child2))

        
# Replace the old population with the new population
population = new_population

# Print the best individual of the current generation
best_individual = max(population, key=fitness_function)
best_fitness = fitness_function(best_individual)
print(f'Generation {0}: Best Fitness = {best_fitness}, Best Individual = {best_individual}')

for generation in range(1,NUM_GENERATIONS):
    # Evaluate fitness of each individual
    fitnesses = [fitness_function(ind) for ind in population] 


    # Sort the population by fitness and preserve the top individuals
    sorted_population = [ind for ind, fit in sorted(zip(population, fitnesses), key=lambda x: x[1], reverse=True)]
    elites = sorted_population[:ELITISM_SIZE]
    elites_fitnesses = [fitness_function(ind) for ind in elites]

    # Create a new population through selection, crossover, and mutation
    new_population = elites  # Start with the best individuals

    while len(new_population) < POPULATION_SIZE:
        parent1 = tournament_selection(elites, elites_fitnesses)
        parent2 = tournament_selection(elites, elites_fitnesses)
        child1, child2 = crossover(parent1, parent2)
        new_population.append(mutate(child1))
        if len(new_population) < POPULATION_SIZE:
            new_population.append(mutate(child2))

        
    # Replace the old population with the new population
    population = new_population

    # Print the best individual of the current generation
    best_individual = max(population, key=fitness_function)
    best_fitness = fitness_function(best_individual)
    print(f'Generation {generation}: Best Fitness = {best_fitness}, Best Individual = {best_individual}')

    
    


    
```
